## TASK 11. Normalization
Normalization changes numeric values to a 0-to-1 range. Following the course example, demonstrate this on MonthlyIncome in a new column so the original values remain available. If every value were identical, division by the maximum would not be useful; inspect the result.

```python
income_max = df['MonthlyIncome'].max()
df['MonthlyIncome_Normalized'] = df['MonthlyIncome'] / income_max
df[['MonthlyIncome', 'MonthlyIncome_Normalized']].head()
```

Check the normalized values with describe().

```python
df['MonthlyIncome_Normalized'].describe().round(3)
```

## TASK 12. Standardization
Standardization expresses numeric values relative to their mean and spread. The guided project demonstrates StandardScaler. Apply it to MonthlyIncome in a separate column, leaving the raw income unchanged.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df['MonthlyIncome_Standardized'] = scaler.fit_transform(
    df['MonthlyIncome'].values.reshape(-1, 1)
)
df[['MonthlyIncome', 'MonthlyIncome_Standardized']].head()
```

A quick check: standardized values should be centered around zero. Small rounding differences are normal.

```python
df['MonthlyIncome_Standardized'].describe().round(3)
```

## TASK 13. Binning
Binning groups numeric values into ranges. The next cell divides MonthlyIncome into three equal-width ranges. The range counts may be uneven. The labels are examples for practicing pandas; they are not official salary categories or business thresholds.

```python
income_bins = np.linspace(min(df['MonthlyIncome']), max(df['MonthlyIncome']), 4)
income_labels = ['Bin 1', 'Bin 2', 'Bin 3']
df['MonthlyIncome_Band'] = pd.cut(
    df['MonthlyIncome'], bins=income_bins, labels=income_labels,
    include_lowest=True
)
df['MonthlyIncome_Band'].value_counts(sort=False) 
```

## TASK 14. Apply a simple function with pandas

MonthlyIncome is recorded as a monthly amount. For practice, apply a function that multiplies it by 12 to estimate annual income, assuming the monthly amount stays the same for all 12 months. This is a demonstration and does not change the original MonthlyIncome column.

```python
def monthly_to_annual(monthly_income):
    return monthly_income * 12

df['AnnualIncome_Demo'] = df['MonthlyIncome'].apply(monthly_to_annual)

df[['MonthlyIncome', 'AnnualIncome_Demo']].head()
```
